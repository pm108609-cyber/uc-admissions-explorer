```python
import pandas as pd

# 1. Load the freshman discipline file
df_discipline = pd.read_csv('uc_freshman_admission_by_discipline.csv')

# 2. Let's peek at the columns and data rows to find the exact labels
print("--- DISCIPLINE DATA COLUMNS ---")
print(df_discipline.columns)

print("\n--- FIRST 5 ROWS ---")
df_discipline.head()
```

```python
# 1. Filter ethically to focus only on the requested Fall 2025 cycle
df_disc_2025 = df_discipline[df_discipline['fall_term'] == 2025]

# 2. Extract the overall campus baseline admit rates (All disciplines)
overall_rates = df_disc_2025[df_disc_2025['broad_discipline'] == 'All disciplines'][['campus', 'admit_rate']]
overall_rates = overall_rates.rename(columns={'admit_rate': 'overall_rate'})

# 3. Extract the specific Computer Science admit rates
cs_rates = df_disc_2025[df_disc_2025['broad_discipline'] == 'Computer Science'][['campus', 'admit_rate']]
cs_rates = cs_rates.rename(columns={'admit_rate': 'cs_rate'})

# 4. Merge the two slices together to compare them side-by-side
comparison_df = pd.merge(overall_rates, cs_rates, on='campus')

# 5. Calculate the "Admit Rate Cost" (The drop in acceptance rate for CS)
comparison_df['admit_rate_cost'] = comparison_df['overall_rate'] - comparison_df['cs_rate']

# 6. Sort the table to see which campus has the biggest gap at the top
sorted_comparison = comparison_df.sort_values(by='admit_rate_cost', ascending=False)

print("--- ADMIT RATE GAP ANALYSIS (FALL 2025) ---")
print(sorted_comparison.to_string(index=False))
```

```python
import pandas as pd

# 1. Load the ethnicity data file
df_eth = pd.read_csv('uc_admissions_summary_by_ethnicity.csv')

# 2. Print out columns and unique ethnic labels to map our filters
print("--- ETHNICITY DATA COLUMNS ---")
print(df_eth.columns)

# Look for the demographic categories column (usually named 'ethnicity' or similar)
eth_col = [col for col in df_eth.columns if 'ethn' in col.lower() or 'race' in col.lower()][0]
print(f"\nUnique labels inside '{eth_col}':")
print(df_eth[eth_col].unique())
```

```python
# Check the exact text labels used inside the count_type column
print("--- Unique values in 'count_type' ---")
print(df_eth['count_type'].unique())
```

```python
# 1. Filter for Fall 2025, remove summary rows, and keep ONLY Freshman level entries
df_2025 = df_eth[
    (df_eth['fall_term'] == 2025) &
    (df_eth['campus'] != 'Universitywide') &
    (df_eth['entrant_level'].str.lower().str.contains('freshman|undergrad', na=False))
]

# 2. Pivot the table cleanly using our new abbreviated column labels
pivoted = df_2025.pivot_table(index=['campus', 'ethnicity'], columns='count_type', values='n', aggfunc='sum').reset_index()

# 3. Calculate the admit rate for each campus and ethnicity group using 'Adm' and 'App'
pivoted['admit_rate'] = pivoted['Adm'] / pivoted['App']

# 4. Separate the table into White and Hispanic/Latino(a) subsets to compare them
white_rates = pivoted[pivoted['ethnicity'] == 'White'][['campus', 'admit_rate']].rename(columns={'admit_rate': 'white_rate'})
hispanic_rates = pivoted[pivoted['ethnicity'] == 'Hispanic/Latino(a)'][['campus', 'admit_rate']].rename(columns={'admit_rate': 'hispanic_rate'})

# 5. Merge the subsets together side-by-side
comparison = pd.merge(white_rates, hispanic_rates, on='campus')

# 6. Flag where the White admit rate is strictly higher
comparison['white_higher'] = comparison['white_rate'] > comparison['hispanic_rate']

# Print the final detailed breakdown table
print("--- ADMIT RATE COMPARISON (FALL 2025) ---")
print(comparison.to_string(index=False))

# Count how many campuses meet the criteria
higher_count = comparison['white_higher'].sum()
print("\n--- FINAL ANSWER ---")
print(f"Number of UC campuses where White freshman admit rate was higher: {higher_count}")
```

```python
# 1. Filter for Fall 2025 in the discipline dataset
df_2025_raw = df_discipline[df_discipline['fall_term'] == 2025]

# 2. Sum up raw applicants and admits to see if pre-calculated columns are rounded or skewed
raw_summary = df_2025_raw.groupby(['campus', 'broad_discipline']).agg(
    total_apps=('applicants', 'sum'),
    total_adms=('admits', 'sum')
).reset_index()

# 3. Calculate the precise raw admit rate
raw_summary['precise_rate'] = raw_summary['total_adms'] / raw_summary['total_apps']

# 4. Separate overall and CS to find the exact relative drop
overall_raw = raw_summary[raw_summary['broad_discipline'] == 'All disciplines'][['campus', 'precise_rate']].rename(columns={'precise_rate': 'overall_rate'})
cs_raw = raw_summary[raw_summary['broad_discipline'] == 'Computer Science'][['campus', 'precise_rate']].rename(columns={'precise_rate': 'cs_rate'})

raw_comparison = pd.merge(overall_raw, cs_raw, on='campus')
raw_comparison['relative_drop'] = ((raw_comparison['overall_rate'] - raw_comparison['cs_rate']) / raw_comparison['overall_rate']) * 100

print("--- RAW COUNTS RELATIVE DROP RANKING ---")
print(raw_comparison.sort_values(by='relative_drop', ascending=False).to_string(index=False))
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import plotly.express as px

# 1. Title and project motivation
st.set_page_config(layout="wide")
st.title("🎓 UC Admissions Equity & Representation Explorer")
st.markdown("""
This data tool evaluates **equity in action** across the University of California system.
It analyzes how freshman acceptance rates vary across different ethnic demographics to identify
where representation gaps exist for the **Fall 2025** cohort.
""")

# 2. Load the ethnicity dataset safely
@st.cache_data
def load_data():
    df = pd.read_csv('uc_admissions_summary_by_ethnicity.csv')
    # Filter out summary rows and focus strictly on Freshman entry metrics
    df = df[(df['campus'] != 'Universitywide') & (df['entrant_level'].str.lower().str.contains('freshman', na=False))]
    # Pivot our abbreviated App and Adm counters cleanly into metrics
    pivoted = df.pivot_table(index=['fall_term', 'campus', 'ethnicity'], columns='count_type', values='n', aggfunc='sum').reset_index()
    pivoted['Admit Rate (%)'] = (pivoted['Adm'] / pivoted['App']) * 100
    pivoted = pivoted.rename(columns={'App': 'Total Applicants', 'Adm': 'Total Admits'})
    return pivoted

try:
    df_clean = load_data()

    # 3. Sidebar Interactivity Settings
    st.sidebar.header("🗺️ Filter Panel")

    # Filter by Year
    years = sorted(df_clean['fall_term'].unique(), reverse=True)
    selected_year = st.sidebar.selectbox("Select Cohort Year:", years)
    df_filtered_year = df_clean[df_clean['fall_term'] == selected_year]

    # Filter by Campus
    campuses = sorted(df_filtered_year['campus'].unique())
    selected_campus = st.sidebar.selectbox("Select UC Campus:", campuses)

    # Final localized data slice
    final_data = df_filtered_year[df_filtered_year['campus'] == selected_campus].sort_values(by='Admit Rate (%)', ascending=False)

    # 4. Dashboard Metrics Layout
    st.subheader(f"📊 Admission Insights for {selected_campus} ({selected_year})")

    total_apps = int(final_data['Total Applicants'].sum())
    total_adms = int(final_data['Total Admits'].sum())
    overall_rate = (total_adms / total_apps) * 100 if total_apps > 0 else 0

    kpi1, kpi2, kpi3 = st.columns(3)
    kpi1.metric("Total Pool Applicants", f"{total_apps:,}")
    kpi2.metric("Total Pool Admits", f"{total_adms:,}")
    kpi3.metric("Campus Baseline Admit Rate", f"{overall_rate:.2f}%")

    st.markdown("---")

    # 5. Interactive Data Visualization Layout
    col_chart, col_table = st.columns([2, 1])

    with col_chart:
        st.markdown("#### **Admit Rates by Ethnic Demographic**")
        fig = px.bar(
            final_data,
            x='ethnicity',
            y='Admit Rate (%)',
            text=final_data['Admit Rate (%)'].apply(lambda x: f'{x:.1f}%'),
            color='ethnicity',
            color_discrete_sequence=px.colors.qualitative.Pastel
        )
        fig.update_layout(showlegend=False, xaxis_title="Demographic Group", yaxis_title="Acceptance Percentage (%)")
        st.plotly_chart(fig, use_container_width=True)

    with col_table:
        st.markdown("#### **Raw Data Audit View**")
        st.dataframe(
            final_data[['ethnicity', 'Total Applicants', 'Total Admits', 'Admit Rate (%)']],
            use_container_width=True,
            hide_index=True
        )

except Exception as e:
    st.error(f"Error compiling app dataset: {e}")
    st.info("Make sure 'uc_admissions_summary_by_ethnicity.csv' is uploaded to your left side storage panel!")
```
